## Challenge 1: Predict the Output

**Objective:** Predict the measurement outcomes of three simple quantum circuits, execute each circuit using Qiskit for 100 shots, and compare the predictions with the actual results.

**Tools:** Python, Qiskit, Qiskit Aer, Matplotlib.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()
```

## 1. Theory

A qubit is the basic unit of quantum information. A new qubit in Qiskit starts in the state |0⟩.

**X gate:** Flips the qubit state. It changes |0⟩ to |1⟩ and |1⟩ to |0⟩.

**H gate:** Creates a superposition. When applied to |0⟩, it gives an equal probability of measuring 0 or 1.

**Measurement:** Converts the quantum state into a classical result, either 0 or 1.

**Shot:** One execution of a circuit. Each circuit in this challenge will run for 100 shots.

```python
circuit_a = QuantumCircuit(1, 1)

circuit_a.x(0)
circuit_a.measure(0, 0)

result_a = simulator.run(circuit_a, shots=100).result()
counts_a = result_a.get_counts()

print("Measurement results:", counts_a)
fig = plot_histogram(counts_a)
display(fig)
plt.close(fig)
```

### Circuit A — Prediction and Observation

**Prediction:** All 100 measurements will produce 1.

**Reason:** The X gate flips the initial state from |0⟩ to |1⟩.

**Actual result:** Measurement results: {'1': 100}.

**Comparison:** Prediction matched.

```python
circuit_b = QuantumCircuit(1, 1)

circuit_b.x(0)
circuit_b.x(0)
circuit_b.measure(0, 0)

result_b = simulator.run(circuit_b, shots=100).result()
counts_b = result_b.get_counts()

print("Measurement results:", counts_b)
fig = plot_histogram(counts_b)
display(fig)
plt.close(fig)
```

### Circuit B — Prediction and Observation

**Prediction:** All 100 measurements will produce 0.

**Reason:** The first X gate changes |0⟩ to |1⟩. The second X gate changes |1⟩ back to |0⟩.

**Actual result:** Measurement results: {'0': 100}

**Comparison:** Prediction .

```python
circuit_c = QuantumCircuit(1, 1)

circuit_c.h(0)
circuit_c.measure(0, 0)

result_c = simulator.run(circuit_c, shots=100).result()
counts_c = result_c.get_counts()

print("Measurement results:", counts_c)

fig = plot_histogram(counts_c)
display(fig)
plt.close(fig)
```

### Circuit C — Prediction and Observation

**Prediction:** Approximately 50 measurements will produce 0 and 50 will produce 1.

**Reason:** The H gate creates an equal superposition of |0⟩ and |1⟩. Each outcome has a 50% probability of being measured.

**Actual result:** Measurement results: {'0': 51, '1': 49}

**Comparison:** The results should be approximately balanced, but they do not have to be exactly 50–50.
