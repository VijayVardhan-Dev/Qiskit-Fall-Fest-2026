## Challenge 2: Control the Measurement Probabilities

**Objective:** Create a single-qubit circuit with unequal probabilities of measuring 0 and 1 using quantum gates.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as pltb
import numpy as np

simulator = AerSimulator()
```

## Theory and Prediction

A Hadamard (H) gate creates an equal superposition of |0⟩ and |1⟩, giving each outcome a 50% probability.

An Ry gate rotates the qubit by an angle θ. Starting from |0⟩, it creates the state:

cos(θ/2)|0⟩ + sin(θ/2)|1⟩

Therefore:

- Probability of 0 = cos²(θ/2)
- Probability of 1 = sin²(θ/2)

By changing θ, we can control the measurement probabilities.


```python
circuit_a = QuantumCircuit(1,1)

theta = np.pi / 3
circuit_a.ry(theta,0)
circuit_a.measure(0,0)

result_a = simulator.run(circuit_a , shots = 100).result()
count_a = result_a.get_counts()
fig = plot_histogram(count_a)
display(fig)
plt.close(fig)
```

```python
zeros = count_a.get('0', 0)
ones = count_a.get('1', 0)

print("Predicted probabilities:")
print("0:", 75, "%")
print("1:", 25, "%")

print("\nActual measurement results:")
print("0:", zeros, "shots =", zeros, "%")
print("1:", ones, "shots =", ones, "%")
```
